# 01 — Exploratory Data Analysis (EDA)

Inspect and explore the raw & processed datasets for SmartHire:
- Resume Dataset (`UpdatedResumeDataSet.csv`): Categories, text distributions, sample resumes.
- Job Postings Dataset (`jobs_clean.csv` / `job_corpus.csv`): Job titles, companies, skills, locations.

```python
import sys
from pathlib import Path

# Add project root to path
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src import config

print(f"Project root: {PROJECT_ROOT}")
```

## 1. Explore Resume Dataset

```python
# Load cleaned resumes
resumes_df = pd.read_csv(config.RESUMES_CLEAN_CSV)
print(f"Resumes shape: {resumes_df.shape}")
print(f"Unique categories: {resumes_df['Category'].nunique()}")
resumes_df.head()
```

```python
# Category distribution
plt.figure(figsize=(12, 6))
cat_counts = resumes_df['Category'].value_counts()
sns.barplot(x=cat_counts.values, y=cat_counts.index, palette="viridis")
plt.title("Distribution of Resume Categories")
plt.xlabel("Number of Resumes")
plt.show()
```

## 2. Explore Job Postings Corpus

```python
jobs_df = pd.read_csv(config.JOBS_CLEAN_CSV)
print(f"Job corpus shape: {jobs_df.shape}")
jobs_df.head(3)
```

```python
# Top job sources and missing values summary
print("=== Source Distribution ===")
print(jobs_df['source'].value_counts())

print("\n=== Missing Values Summary ===")
print(jobs_df.isnull().sum())
```
