# 03 — Unsupervised Job Recommender

Content-based job recommendation system using TF-IDF + Cosine Similarity.
Given a candidate resume, computes vector similarity against 136k+ job postings and returns top-N ranked matches.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from src.models.recommender import load_recommender, recommend_jobs
from src.features.match_features import extract_skills, recommend_skills_to_learn

job_vec, job_matrix, job_meta = load_recommender()
print(f"Catalog loaded: {len(job_meta):,} jobs | Matrix shape: {job_matrix.shape}")
```

## Generate Recommendations for Sample Resume

```python
sample_text = """
Senior Java Engineer with 5 years experience in Spring Boot, REST APIs, Microservices,
Docker, PostgreSQL, AWS, and React.
"""
recs = recommend_jobs(sample_text, top_n=5, job_vectorizer=job_vec, job_matrix=job_matrix, job_metadata=job_meta)

for r in recs:
    print(f"#{r['rank']} {r['title']} @ {r['company']} (Match Score: {r['match_score']}%)")
    print(f"   Matched: {r['matched_skills']}")
    print(f"   Missing: {r['missing_skills']}\n")
```
